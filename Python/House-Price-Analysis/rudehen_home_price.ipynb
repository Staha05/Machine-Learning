```python
X = df.loc[:,(df.columns != 'قیمت کل') & (df.columns != 'قیمت هر متر')]
Y = df['قیمت کل']
```

```python
print(f"تعداد ویژگی‌های X: {X.shape}")
print(f"نام ستون‌های X: {X.columns.tolist()}")
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.40, random_state = 1)
X_train.shape
```

```python
print(f"X shape{X_test.shape}")
print(f"Y shape{Y_test.shape}")
```

```python
model = LinearRegression()
```

```python
model.fit(X_train, Y_train)
```

```python
features_for_prediction = [1, 5000, 1404, 3, 5, 0, 0, 0, 0, 0, 0]
```

```python
features_df = pd.DataFrame([features_for_prediction], columns=X_train.columns)

predicted_total_price = model.predict(features_df)
print(f"قیمت کل پیش‌بینی شده: {predicted_total_price[0]:,.0f} تومان")
```
